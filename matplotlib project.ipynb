```python
import matplotlib.pyplot as plt
```

```python
#data
x=[1,2,3,4,5]
y=[10,20,25,30,45]
plt.plot(x,y)
plt.show()
plt.figure(figsize=(2,2))
```

```python
plt.plot(x, y, color='r', marker='o', linestyle='dashed',linewidth=2, markersize=12)
plt.title("chart")
plt.xlabel("x axis ka label hai")
plt.ylabel("y axis ka label hai")
plt.figure(figsize=(4,3))
plt.show()
```

```python
#advanced 
```

```python
x=[1,2,3,4,5]
y1=[10,20,25,30,45]
y2=[20,30,35,45,55]
plt.plot(x,y1, label="sales 2024")#y1 data
plt.plot(x,y2, label="sales 2025")#2 data
plt.title("year sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.legend()
plt.show()
```

```python
x=["A","B","C","D","E"]
y=[10,20,25,30,45]
plt.bar(x,y)
plt.title("bar chart by rishi")
plt.show()
```

```python
##Histogram chart
```

```python
import random
data=[random.randint(1,10)for _ in range(100)]
plt.hist(data,bins=5)
plt.title("histogram chart")
plt.show()
```

```python
## Pie charts
x=["A","B","C","D","E"]
y=[10,20,25,30,45]
plt.pie(y,labels=x,autopct= "%1.1f%%",startangle=90)
plt.title("pie chart")
plt.show()
```

```python
## scatter plot
```

```python
## Used to find the values
y1=[10,20,25,30,45]
y2=[20,30,35,45,55]
plt.scatter(y1,y2)
plt.title("Scatter plot chart")
plt.show()
```

```python
##used multiple chart in a single chart
```

```python
##data 1 --bar chart
x=["mon","tue","wed","thu","fri"]
y=[10,20,25,30,45]
## scatter chart
y1=[10,20,25,30,45]
y2=[20,30,35,45,55]
plt.figure(figsize=(6,4))
## create subplot chart
plt.subplot(1,1,1) #row columns,position
plt.bar(x,y)
plt.title("daily sales")
plt.xlabel("weak days")
plt.ylabel("sales")
plt.show()
## scatter plot------

plt.subplot(1,1,1) #row columns,position
plt.scatter(y1,y2)
plt.title(" scatter user")
plt.xlabel("user1")
plt.ylabel("user2")
plt.show()
```

```python
## matplotlib with pandas---
```

```python
import pandas as pd
#data
data= {
    "month":["jan","feb","mar","apr"],
    "sales":[12000,11000,13000,25000]
    }
df=pd.DataFrame(data)
df
```

```python
plt.bar(df["month"],df["sales"])
plt.title("matplotlib with pandas")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```

```python
## save chart pd
```

```python
plt.bar(df["month"],df["sales"])
plt.title("matplotlib with pandas")
plt.xlabel("month")
plt.ylabel("sales")
plt.savefig("monthly_sales_from_df.png")
plt.show()
```
